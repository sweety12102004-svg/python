```python
employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]

new_list = [i["name"] for i in employees if i["salary"] > 50000]

print(new_list)
```

```python
transactions = [5000, -2000, 3000, -1000, 7000, -500]

deposits = [i for i in transactions if i > 0]

print(deposits)
withdrawals = [i for i in transactions if i < 0]

print(withdrawals)
```
