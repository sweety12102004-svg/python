```python
balance=3000  #Global Variable
def show():  # Function without parameter
    deposit=1000
    print("Deposit Balance: ", deposit)
    print("Initial Balance was: ", globals() ["balance"]) #One time use

show()
print("Current Balance After Deposit: ", balance)
```

```python
x=30  # Global Variable
def show():
    global x
    print("Global Variable: ",x)  # 30
show()
print("Global Variable: ",x)  # 30

def display():
    global x
    print("Global Variable: ",x)  # 30
display()
```

```python
balance=3000  # Global Variable
def show():
    global balance
    deposit=1000
    print("Deposit Balance: ",deposit)

    balance=balance+deposit  # 4000

show()
print("Current Balance After Deposit: ",balance)
```

```python
def CalculateBill(price,discount=0):
    discount_amount=price*discount/100
    bill=price-discount_amount

    print("Original Price: ",price)
    print("Discount: ",discount, "%")
    print("Discount Amount: ",discount_amount)
    print("Final Bill Amount: ",bill)
    print("============================")

CalculateBill(20000)  # Cash 0% discount
CalculateBill(10000,10)  # UPI 10% discount
CalculateBill(30000,20)  # Card 20% discount
```

```python
def add(a,b):
    return a+b
def add(a,b,c):
    return a+b+c
def add(a,b,c,d):
    return a+b+c+d
def add(a,b,c,d,e):
    return a+b+c+d+e

add(10,20) # 2
add(20,30,40) # 3
add(20,30,40,50) # 4
add(20,30,40,50,60) # 5
```

```python
# args
def add(*args):  # Tuple
    total=0  # 30
    for ex in args:
        total=total+ex  # 150
    print("Sum: ",total)
add(10,20)  # args=(10,20)
add(40,50,60) # args=(40,50,60)
add(20,30,40,50)  # args=(20,30,40,50)
add(20,30,40,50,60) # args=(20,30,40,50,60)
```

```python
def add(*args):
    print("args[0]",args[0]) # indexing
    print("args[1]",args[1]) # indexing
    print("args[2]",args[2]) # indexing
    print("args[3]",args[3]) # indexing
    print("args[4]",args[4]) # indexing
    print(args)  # (20,30,40,50,60) Tuple
                 # [20,30,40,50,60] List

add(20,30,40,50,60)  # args=(20,30,40,50,60)
```

```python
# *kwargs # Keyword Arguments >> Dictionary

def EmployeeDetails(**kwargs):
    print(kwargs)

EmployeeDetails(name="Mohit",Role="Developer",Department="IT")
EmployeeDetails(name="Mohit",Role="Developer")
EmployeeDetails(name="Mohit")

#EmployeeDetails("Mohit","Developer","IT")
#kwargs={name;"Mohit",Role:"Developer",Deaprtment:"IT"}
#kwargs={name;"Mohit",Deaprtment:"IT"}
#kwargs={name;"Mohit"}
```

```python
def Employeedetails(**kwargs):
    print(kwargs['name'])
    print(kwargs['Role'])
    print(kwargs['Department'])

Employeedetails(name="Mohit",Role="Developer",Department="IT")
```

```python
def Employeedetails(**kwargs):
    for key, value in kwargs.items():
        print(key, ";", value)

    print("=============================")

Employeedetails(name="Mohit",Role="Developer",Department="IT")
Employeedetails(name="Mohit",Department="IT")
Employeedetails(name="Mohit")
```

```python
#Exercise:4
def studentResult(name,*marks,**details):
    print("Student Name: ",name)
    print("Subject Marks: ",marks)

    if len(marks)>0: #false
        total=sum(marks)
        avg=total/len(marks)
        print("total Marks: ",total)
        print("Avg Marks: ",avg)

    print("=============Additional Information==========")
    for key, value in details.items():
        print(key, ":", value)

studentResult("Amit", 50,89,98,78,79, roll=101, Sem="1st")
```

```python
def addItems(*prices):
    print("Prices: ",prices)
    print("Total: ",sum(prices))

def customerDetails(Name, City, Member):
    print("Name: ",Name)
    print("City: ",City)
    print("Member: ",Member)

prices=[20,65,77,85,35]
addItems(*prices)

customer={
    "Name":"Aman",
    "City":"Delhi",
    "Member":"Gold"
}

customerDetails(**customer)
```

```python
def calculate_avg(*marks):
    if not marks:
        return "Error: No Marks Provided"
    for ex in marks:
        if not isinstance(ex,(int,float)) or isinstance(ex, bool):
            return "Error: Marks Must be Numaric"
        if ex<0 or ex>100:
            return "Error: Marks must be between 0 and 100"
    return sum(marks)/len(marks)

print(calculate_avg(55,45,58,97.0,87,67))
print(calculate_avg())
print(calculate_avg(47,87,True))
```
